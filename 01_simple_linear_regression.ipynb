## How Linear Regression Works

### What is linear regression?

In statistics, **linear regression** is an approach for modelling the relationship between a response variable $y$ and one or more explanatory variables. Keep in mind that linear regression can approximate a relationship, but it cannot prove causality.

Linear regression also makes some **assumptions**:
1. **Linearity**: The model is linear in its coefficients: the target is a weighted sum of the explanatory variables, plus an error.
2. **Zero-Mean Error**: On average, the errors (the part of $y$ that the line cannot explain) are zero.
3. **Strict Exogeneity**: The errors are uncorrelated with all the explanatory variables.
4. **Homoscedasticity**: The variance of the errors is constant across all levels of the explanatory variables.
5. **No Multicollinearity**: All the explanatory variables are linearly independent.

We can divide linear regression into two categories:
* **Simple linear regression**: cases in which we have only one explanatory variable.
* **Multiple linear regression**: cases in which we have more than one explanatory variable.

The formula for a simple linear regression is

$$
y_i = b_0 + b_1 x_i + e_i
$$  
$$
\text{for i = } 1, \dots, n
$$

where $y_i$ is the true value of the response variable and $e_i$ is the error. We never observe the true errors: the **residuals** of a fitted model are our estimates of them, which is why the assumptions are checked on the residuals. The formula for the regression equation is

$$
\hat{y}_i = b_0 + b_1 x_i
$$ 
$$
\text{for i = } 1, \dots, n
$$

where $\hat{y}_i$ is the predicted value of the response variable, and $b_0$ and $b_1$ are the estimated intercept and slope of the line.

> **The Gauss-Markov theorem.** When these assumptions hold (and, in addition, the errors are uncorrelated with each other), the ordinary least squares (OLS) estimates of $b_0$ and $b_1$ are the **best linear unbiased estimators (BLUE)**. Of all the linear, unbiased ways to estimate the coefficients, OLS has the smallest variance.
>
> Note what the theorem does *not* promise: it does not require normally distributed residuals, and it does not guarantee good predictions when the assumptions are broken. The optional notebook 07 (Checking the Assumptions) shows how to check each assumption on real data.

### How does it work?



Linear regression finds the **line of best fit**. In other words, we look for the **intercept** and **slope** that define a line that fits the data as well as possible.

"As well as possible" usually means minimising the sum of squared **residuals**. This approach is called the **Ordinary Least Squares (OLS)** method.

* **Intercept**: The value of our response variable $y$ when our explanatory variable $x=0$ ($b_0$ in the formula above).
* **Slope**: For each unit increase in $x$, the expected increase or decrease in $y$ ($b_1$ in the formula above). This definition is sufficient only for simple linear regression. In multiple linear regression there is more than one explanatory variable in the model, so we have to add "holding all other explanatory variables constant" to the definition.
* **Residual**: The difference between an observed value and the fitted value provided by a model: $e_i = y_i - \hat{y}_i$
* **Ordinary Least Squares (OLS)**: OLS is one of the simplest and most common methods used to find the parameters of a linear regression model. It minimises the sum of squared residuals: $\sum_{i=1}^n e_i^2$. You can find a visual explanation in this [interactive OLS explainer](https://setosa.io/ev/ordinary-least-squares-regression/).

## Simple linear regression with scikit-learn



In this section we will train our first linear regression model with the scikit-learn library.

So far, we have talked about response and explanatory variables. In data science, it is far more common to call the response variable, the one we want to predict, the **target variable**. The explanatory variables are usually called **features**, and the slopes are often called regression **coefficients**.

Scikit-learn makes its algorithms easy to use, as long as we organise our data in the format its models expect. The workflow therefore has three steps:
- First, define the target variable (`mpg`) and the feature (`weight`).
- Then, instantiate a [`LinearRegression`](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LinearRegression.html) object.
- Finally, train it by calling its `.fit` method with our data.

Note that miles per gallon (`mpg`) measures distance per unit of fuel. It is the inverse of the litres per 100 km used in Germany: a higher `mpg` means a more economical car.

```python
# Import linear regression from sklearn library
from sklearn.linear_model import LinearRegression
```

```python
# Prepare data for modeling by defining target and feature
X = cars[["weight"]]  # X needs to be 2-dimensional so we need double brackets here
y = cars["mpg"]
```

```python
# Show dimensions of X
X.shape
```

```python
# Instantiate our model
lin_reg = LinearRegression()

# Train the model using our data
lin_reg.fit(X, y)
```

```python
# Intercept and slope/coefficient can be accessed via .intercept_ and .coef_
intercept = lin_reg.intercept_
slope = lin_reg.coef_[0]
print("Model intercept:", intercept)
print("Coefficient for feature weight:", slope)
```

```python
# Plotting our data points
plt.scatter(X, y)

# Adding the fitted regression line of our model
plt.plot(X, X * slope + intercept, "-", color="magenta", linewidth=2)

plt.title("Relationship between mpg and weight")
plt.ylabel("mpg")
plt.xlabel("weight");
```

In the cells above we trained the linear regression model on our data. With the values of the intercept and the coefficient, we added the regression line to the scatter plot.

We can also interpret the intercept of our model: according to the model, a car with a `weight` of 0 would have an `mpg` of about 37.28. As this case shows, an intercept may or may not have a meaning in the real world: a car cannot have a weight of 0.

The value of the coefficient tells us that we expect `mpg` to decrease by about 5.34 when `weight` goes up by 1.

### Quantifying the fit of the model




So far we have seen how to estimate the coefficients of a simple linear regression model. What we are still missing is a proper way to evaluate whether our model did a good job or a poor one.

With only one feature, we can plot the regression line over the data to get a feeling for the fit. As soon as we add more features, however, there are too many dimensions to plot. Plotting and eyeballing is also an inaccurate way to measure the fit.

#### $R^2$ (r-squared)



We can assess the fit of our model numerically by calculating its **$R^2$**. It has a simple interpretation: the $R^2$ is the *proportion* of the variance in the target variable (in our case `mpg`) that the features account for (in our case there is only one feature, `weight`).

Mathematically, we define $R^2$ as

$$R^2 = 1 - \frac{SSR}{SST}$$

where **SSR** is the sum of squared residuals ($\sum_{i=1}^n e_i^2$) and **SST** is the total sum of squares ($\sum_{i=1}^n (y_i - \bar{y})^2$).

On the data the model was trained on (for a model with an intercept), $R^2$ lies between 0 and 1. On new data it can even be negative, when the model predicts worse than simply using the mean of $y$; you will meet this once models are evaluated on test sets. If the model captures the data perfectly, the **SSR** is 0 and $R^2$ is 1. A poor fit, on the other hand, results in a high **SSR**, and $R^2$ is low, close to 0.

#### $R^2$ in scikit-learn


We don't have to calculate the $R^2$ by hand. It is already implemented in scikit-learn, and we can import it from the `sklearn.metrics` module. The [scikit-learn documentation](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.r2_score.html) for the `r2_score` function shows that you have to pass at least two arguments, `y_true` and `y_pred`:
* `y_true` are the true values of our target variable (in our case stored in the variable `y`).
* `y_pred` are the estimated target values. We called them $\hat{y}$ at the beginning of the notebook. We could calculate them from the slope and the intercept, but scikit-learn provides a more convenient method called `.predict`.

```python
# Import r-squared from sklearn
from sklearn.metrics import r2_score
```

```python
# Calculate the estimated value for each data point
y_hat = lin_reg.predict(X)

# Calculate the R-squared for our model
print("R-squared:", round(r2_score(y, y_hat), 3))
```

An $R^2$ of 0.753 looks good at first, but you should not rely on that measure alone when you assess your model. An $R^2$ of 0.753 "only" means that our feature `weight` explains 75.3 % of the variance in our target variable `mpg`.

A later session covers how to evaluate a model's performance properly.

### Plotting the Residuals

$R^2$ is a single number, so it cannot show *where* the model goes wrong. The residuals can. Recall that a residual is the gap between the true and the predicted value, $e_i = y_i - \hat{y}_i$.

Plot the residuals against the predicted values. If the model fits well, the points form a random band around zero, with no pattern.

```python
# Residuals: the true mpg minus the mpg our model predicts
residuals = y - y_hat

plt.scatter(y_hat, residuals)
plt.axhline(0, color="magenta", linewidth=2)
plt.title("Residuals of the model")
plt.xlabel("predicted mpg")
plt.ylabel("residual");
```

The residuals are not quite a random band. At both ends, for the heaviest cars (predicted `mpg` below 10) and the lightest cars (above 25), they are mostly positive, and in the middle they are mostly negative. The model underestimates the fuel economy of the heaviest and the lightest cars, a hint that the relationship between `weight` and `mpg` is slightly curved rather than straight.

With only 32 cars, this is a hint, not proof. The optional notebook 07 (Checking the Assumptions) repeats this check on 392 cars, where the curve is clear, and shows how to fix it.

### Predicting for new instances



One of the main reasons we train models in data science is to make predictions for new observations.

In the section above we already used the `.predict` method to calculate the estimated `mpg` values for the data we used to train the model. That allowed us to calculate the $R^2$ of the model.

We can also use `.predict` to estimate a value of the target variable for a new observation. Here, this means calculating the estimated `mpg` of a car with a new weight.

To avoid errors or warnings, the new data point must have the same format as the training data. We therefore define a small DataFrame with one column called `weight` and one row.

```python
# Predict the mpg value for a new car with a weight of 4.5
new_car = pd.DataFrame({"weight": [4.5]})
y_hat_new = lin_reg.predict(new_car)
```

```python
print("Predicted value for new observation:", y_hat_new[0].round(3))
```

According to our model, a car with a weight of 4.5 (4,500 lb) would have a fuel efficiency of 13.235 mpg.
