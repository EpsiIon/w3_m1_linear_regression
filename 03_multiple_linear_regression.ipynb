# Multiple Linear Regression



So far you have worked with simple linear regression, which uses a single explanatory variable. Linear regression can also use more than one explanatory variable to describe the target variable. This is called **multiple linear regression**.

## Recap



### The Multiple Regression Model

 

Multiple linear regression works like simple linear regression, except that the dependent variable $y$ is described by $k$ independent variables $x_1, \dots, x_k$:

$$
y = b_0 + b_1 x_1 + b_2 x_2 + \dots + b_k x_k + e
$$ 

The formula for our **predicted model** is:

$$\hat{y} = b_0  + b_1  x_1 + b_2  x_2 + \dots + b_k  x_k$$

where $b_0$ is the estimated intercept and $b_1, b_2,$... are the coefficients of the features.

* **Intercept**: The interpretation is the same as for simple linear regression. It is the value of $y$ when all $x$ are 0.
* **Coefficients**: Their interpretation needs more care. $b_i$ is the change in $y$ for a one-unit change in $x_i$ while **holding all other variables constant**.

## Let's start with an example



This example uses a car dataset again. It is larger than the previous one, with more observations and more features, but the aim is the same: predict `mpg` from the characteristics of each car.

```python
# Import packages
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Set figure style and size for the entire notebook
sns.set_style("ticks")
plt.rcParams["figure.figsize"] = (7, 4)
```

```python
# Import dataset
cars = pd.read_csv("data/cars_multivariate.csv")
cars.head()
```

## Simple Linear Regression



As a baseline, start with a simple linear regression model that uses only the feature `horsepower`, and check how well it fits the data.

Documentation: [`LinearRegression`](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LinearRegression.html), [`r2_score`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.r2_score.html).

```python
# Plot relationship between horsepower and mpg
plt.scatter(cars.horsepower, cars.mpg)
plt.ylabel("mpg")
plt.xlabel("horsepower");
```

```python
# Define feature and target variable
X = cars[["horsepower"]]
y = cars["mpg"]

X.shape
```

```python
# Fit linear regression model
lin_reg = LinearRegression()
lin_reg.fit(X, y)

# Calculate r-squared
y_hat = lin_reg.predict(X)
print("R-squared:", round(r2_score(y, y_hat), 3))
```

```python
# Plot data with regression line
plt.scatter(X, y)
plt.plot(X, y_hat, "-", color="orange", linewidth=2)
plt.ylabel("mpg")
plt.xlabel("horsepower");
```

According to $R^2$, the feature `horsepower` alone explains 60.6% of the variance in the target variable `mpg`. This is the baseline that the following models should improve on.

## Multiple Linear Regression  



Instead of using `horsepower` as the only independent variable to predict `mpg`, you can include other independent variables in the model to improve the fit. Next, add `weight` to the model.

Documentation: [`lmplot`](https://seaborn.pydata.org/generated/seaborn.lmplot.html).

```python
# Plot relationship between weight and mpg
# We can use seaborn's .lmplot function to plot our data including a regression line
sns.lmplot(x="weight", y="mpg", data=cars, ci=None);
```

```python
# Define model input X with two features
X2 = cars[["horsepower", "weight"]]
y2 = cars.mpg

X2.shape
```

```python
# Fit linear regression model
lin_reg2 = LinearRegression()
lin_reg2.fit(X2, y2)

# Calculate intercept and coefficient
intercept = lin_reg2.intercept_
coefficients = lin_reg2.coef_
print("Intercept:", intercept.round(4))
print("Coefficients:", coefficients.round(4))
```

```python
# Calculate r-squared
y_hat2 = lin_reg2.predict(X2)
print("R-squared:", round(r2_score(y2, y_hat2), 3))
```

### Model Interpretation



Does this model describe the variance of `mpg` better than the simple linear regression? Yes: this model explains about 70% of the variation in `mpg`.

The multiple regression model is:

$$ \hat{mpg} = 45.6402 - 0.0473 \times horsepower - 0.0058 \times weight $$

The terms of this equation mean the following:

* **Intercept**: the expected `mpg` for a car with a `horsepower` and `weight` of 0 is 45.6402. This value is only theoretical, because no real car has zero horsepower or zero weight.
* **`horsepower` coefficient**: we expect `mpg` to decrease by 0.0473 when `horsepower` increases by 1, **holding `weight` constant**.
* **`weight` coefficient**: we expect `mpg` to decrease by 0.0058 when `weight` increases by 1, **holding `horsepower` constant**.

### Predicting for new cars



What is the predicted `mpg` for a car with 200 `horsepower` and a `weight` of 3500?

$$ 
\hat{mpg} = 45.6402 - 0.0473 (200) - 0.0058 (3500) = 15.88 
$$  


We would expect the car to have an `mpg` of 15.88. (`.predict()` gives 15.9: the small difference comes from rounding the coefficients in the hand calculation.)

The model can also calculate this for you. Call `.predict()` and pass it the feature values as a DataFrame with the same column names as the training data.

```python
# Make predictions for new car with horsepower=200 and weight=3500
new_car = pd.DataFrame({"horsepower": [200], "weight": [3500]})
y_prediction = lin_reg2.predict(new_car)
print("Prediction for new car:", y_prediction[0].round(3))
```

### Explore Multiple Regression



A model with two features can still be drawn. With `horsepower` and `weight` as features, the model is no longer a line but a flat **plane** in three dimensions (`horsepower`, `weight` and `mpg`). With more than two features it becomes a **hyperplane**, which we can no longer draw.

The cell below draws the plane of the model you just trained, together with the cars. Points above the plane are cars the model underestimates (a positive residual), and points below are cars it overestimates.

The plot is interactive: drag to rotate it and scroll to zoom.

Documentation: [`Surface`](https://plotly.com/python-api-reference/generated/plotly.graph_objects.Surface.html), [`Scatter3d`](https://plotly.com/python-api-reference/generated/plotly.graph_objects.Scatter3d.html).

```python
import plotly.graph_objects as go

# Intercept and coefficients of the two-feature model trained above
intercept = lin_reg2.intercept_
coef_horsepower, coef_weight = lin_reg2.coef_

# A grid of horsepower and weight values, and the model's prediction for each grid point
hp_grid, weight_grid = np.meshgrid(
    np.linspace(X2.horsepower.min(), X2.horsepower.max(), 30),
    np.linspace(X2.weight.min(), X2.weight.max(), 30),
)
mpg_plane = intercept + coef_horsepower * hp_grid + coef_weight * weight_grid

# True for cars above the plane (positive residual)
above = (y2 - y_hat2) >= 0

fig = go.Figure(
    [
        go.Surface(
            x=hp_grid, y=weight_grid, z=mpg_plane,
            colorscale="RdBu_r", opacity=0.7, showscale=False, name="model",
        ),
        go.Scatter3d(
            x=X2.horsepower[above], y=X2.weight[above], z=y2[above],
            mode="markers", marker=dict(size=3, color="royalblue"), name="above the plane",
        ),
        go.Scatter3d(
            x=X2.horsepower[~above], y=X2.weight[~above], z=y2[~above],
            mode="markers", marker=dict(size=3, color="black"), name="below the plane",
        ),
    ]
)
fig.update_layout(
    scene=dict(xaxis_title="horsepower", yaxis_title="weight", zaxis_title="mpg"),
    height=600,
    margin=dict(l=0, r=0, t=0, b=0),
)
fig.show()
```

The plane slopes down along both axes because both coefficients are negative (-0.047 for `horsepower` and -0.0058 for `weight`): more powerful and heavier cars use more fuel. Rotate the plot to look along the edge of the plane. The cars scatter on both sides of it. This scatter is the 29% of the variance in `mpg` that this model does not explain ($R^2$ = 0.706).

## $R^2$ vs. Adjusted $R^2$



With multiple linear regression, $R^2$ is not a good metric for evaluating and comparing models. The reason: when you add a new feature, $R^2$ either stays the same or increases, even if the feature has no relationship with the target variable. This creates a strong temptation to keep adding features. That is not a good idea. You want a model that fits well, **but** you want it to be as simple as possible.

This is where **adjusted $R^2$** comes in. Adjusted $R^2$ penalises you for adding features that do not improve the model:

- For a simple linear regression, $R^2$ and adjusted $R^2$ are nearly the same.
- The more non-significant features you add, the larger the gap between the two metrics.
- This makes adjusted $R^2$ a useful metric for comparing models with different numbers of features.

The formula of the adjusted $R^2$ is:
$$
R_a^2 = 1 - \frac{(1 - R^2) (n - 1)}{n - p - 1} 
$$  

where $n$ is the number of observations in the dataset and $p$ is the number of features.

scikit-learn does not provide a built-in function for the adjusted $R^2$, so the next cell defines one from the formula above. It takes the R-squared score of a model and its feature matrix `X`: `len(X)` gives $n$ and the number of columns of `X` gives $p$.

```python
# Define function for calculating adjusted r-squared
def adjusted_r_squared(r_squared, X):
    adjusted_r2 = 1 - ((1 - r_squared) * (len(X) - 1) / (len(X) - X.shape[1] - 1))
    return adjusted_r2
```
